```python
print("python pattern programming circle")
print("------------------------------------------")


circle={(2,4),(2,5),(2,6),(3,3),(3,7),(4,3),(4,7),(5,3),(5,7),(6,4),(6,5),(6,6)}
def pat_circle():
    for i in range(1,10):
        for j in range(1,10):
            if (i,j)in circle:
                print('*',end=' ')
            else:
                print(' ', end= ' ')
        print()
pat_circle()
                
```
